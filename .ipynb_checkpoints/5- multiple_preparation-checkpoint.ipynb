```python
import pandas as pd
```

```python
pathological_data = pd.read_csv('data/deduplicated_pathological.csv')
pathological_data.dropna(inplace=True) #drop all NaN values
```

```python
rec_type = 'phrase'
# rec_type = 'a_n'
# rec_type = 'i_n'
# rec_type = 'u_n'
```

```python
pathological_data.drop(pathological_data[pathological_data["recording"] != rec_type].index, inplace=True)
print(pathological_data['type'].value_counts())
```

```python
pathological_data = pathological_data[
    pathological_data["type"].isin(
        pathological_data["type"].value_counts().nlargest(3).index
    )
]
```

```python
#save to a file
pathological_data.to_csv(f"data/preped/multi-{rec_type}.csv", index=False)
```
