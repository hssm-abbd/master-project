```python
import pandas as pd
```

```python
healthy_data = pd.read_csv('data/healthy.csv')
healthy_data.dropna(inplace=True) #drop all NaN values

pathological_data = pd.read_csv('data/deduplicated_pathological.csv')
pathological_data.dropna(inplace=True) #drop all NaN values
```

## Phrase and normal pitch vowel data

```python
# rec_type = 'phrase'
# rec_type = 'a_n'
# rec_type = 'i_n'
rec_type = 'u_n'
```

```python
#Select only rec_type
healthy = healthy_data[healthy_data['recording'] == rec_type]
pathological = pathological_data[pathological_data['recording'] == rec_type]
print(len(healthy))
print(len(pathological))
```

```python
#remove pathologies with less than 20 recordings
counts = pathological['type'].value_counts().to_dict()
for k, v in counts.items():
    if v < 20:
        #print(k)
        pathological = pathological[pathological['type'] != k]
```

```python
#balance data
pathological['_rank'] = pathological.groupby('type').cumcount()

# Sort by rank so lower-rank items from all groups come first
# This caps frequent values while preserving all rare values
pathological_shrunk = (
    pathological.sort_values('_rank')
    .head(len(healthy))
    .drop(columns=['_rank'])
)

#print(len(pathological_shrunk))
#print(pathological_shrunk['type'].value_counts().to_dict())
```

```python
#save to a file
combined = pd.concat([healthy, pathological_shrunk], ignore_index=True)
combined.to_csv(f"data/preped/{rec_type}.csv", index=False)
```
