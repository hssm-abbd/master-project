```python
import pandas as pd
```

```python
pathological_data = pd.read_csv('data/pathological.csv')
pathological_data.dropna(inplace=True) #drop all NaN values
```

## Detect patients with multiple pathologies

```python
#Detect patients with multiple pathologies
result_dict = pathological_data.groupby('person_id')['type'].apply(set).to_dict()
filtered_dict = {k: v for k, v in result_dict.items() if len(v) > 1} #Filtered to only keep patients present in multiple pathologies

print(f"Number of patients: {len(result_dict)}")
print(f"Number of patients with multiple pathologies: {len(filtered_dict)}")
#print(filtered_dict)
#print("="*50)
```

## Remove them from the data

```python
#Removing duplicates (keeping the element within the most frequent pathology)

#Get pathology recording counts
pathology_recording_count = pathological_data['type'].value_counts()
#print(pathology_recording_count)

#Remove duplicates one by one
for patient in filtered_dict.items():
    #print(patient)
    most_frequent_pathology = max(patient[1], key=lambda item: pathology_recording_count.get(item, 0))
    #print(f'{patient[0]}->{most_frequent_pathology}')
    pathological_data = pathological_data[~((pathological_data['person_id'] == patient[0]) & (pathological_data['type'] != most_frequent_pathology))]
```

## Save to new file

```python
pathological_data.to_csv("data/deduplicated_pathological.csv", index=False)
```
